```python
# HAPERPARAMETER TUNNNING C IS PERARAMETER HIGH REGULIZATION
pipe = Pipeline([
    ('preprocessor', trf1),
    ('scaler', StandardScaler()),
    ('model', SVC(C=10))
])

pipe.fit(X_train, y_train)

pipe.score(X_test, y_test)
```

```python
# HAPERPARAMETER TUNNNING C IS PERARAMETER HIGH GAMMA
pipe = Pipeline([
    ('preprocessor', trf1),
    ('scaler', StandardScaler()),
    ('model', SVC(gamma=2))
])

pipe.fit(X_train, y_train)


pipe.score(X_train, y_train)
```
