```python
# Features (all columns EXCEPT director, lead_actor, and lead_actress)
X = df.drop(
    ['director', 'lead_actor', 'lead_actress'], axis=1
)

# Target/Labels (selecting multiple columns requires double brackets)
y = df['director']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=49)
```

```python
X_train.head()
```

```python
trf1 =  ColumnTransformer([
    ('OdEn_title',OrdinalEncoder(handle_unknown='use_encoded_value',
    unknown_value=-1),[1]),
    ('OdEn_genre',OrdinalEncoder(handle_unknown='use_encoded_value',
    unknown_value=-1),[5]),
    ('OdEn_subgenre',OrdinalEncoder(handle_unknown='use_encoded_value',
    unknown_value=-1),[6]),
    ('OdEn_country',OrdinalEncoder(handle_unknown='use_encoded_value',
    unknown_value=-1),[7]),
    ('OdEn_language',OrdinalEncoder(handle_unknown='use_encoded_value',
    unknown_value=-1),[8]),
    ('OdEn[streaming_platform]',OrdinalEncoder(handle_unknown='use_encoded_value',
    unknown_value=-1),[18]),
],remainder = 'passthrough')
```

```python
#  scaling
from sklearn.preprocessing import MinMaxScaler
from sklearn.compose import ColumnTransformer

trf2 = ColumnTransformer(
    transformers=[
        ('scale', MinMaxScaler(), slice(0, 8))  # Notice the parentheses around the tuple!
    ],
    remainder='passthrough'
)
```

```python
# feature selection
from sklearn.feature_selection import SelectKBest, f_classif

trf3 = SelectKBest(
    score_func=f_classif,
    k=8
)
```

```python
trf4 = KNeighborsClassifier()
```

```python
# CREATE PIPLINE
from sklearn.pipeline import Pipeline,make_pipeline
pipe = Pipeline([
    ('trf1',trf1),
    ('trf2',trf2),
    ('trf3',trf3),
    ('trf4',trf4)


])
```

```python
pipe.fit(X_train,y_train)
```

```python
pipe.score(X_train,y_train)
```

```python
y_pred = pipe.predict(X_train)
```

```python
y_pred
```

```python
pipe.score(X_test,y_test)
```

```python
from sklearn.metrics import accuracy_score

y_pred = pipe.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    pipe,
    X_train,
    y_train,
    cv=5,
    scoring='accuracy'
)

print("Cross Validation Scores:", scores)

print("Mean Accuracy:", scores.mean())
```

```python
from sklearn.metrics import classification_report

# Predict values
y_pred = pipe.predict(X_test)

# Display classification report
print(classification_report(y_test, y_pred))
```

```python


from sklearn.model_selection import GridSearchCV

params = {
    'trf4__n_neighbors': [3, 5, 7, 9, 11, 15, 21],
    'trf4__weights': ['uniform', 'distance'],
    'trf4__metric': ['euclidean', 'manhattan']
}

grid = GridSearchCV(
    pipe,
    params,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

grid.fit(X_train, y_train)

print("Best Parameters:", grid.best_params_)
print("Best CV Accuracy:", grid.best_score_)
print("Test Accuracy:", grid.score(X_test, y_test))



```
